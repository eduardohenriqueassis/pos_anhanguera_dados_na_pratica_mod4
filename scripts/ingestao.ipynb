```python
import duckdb
import pandas as pd
import os 
from datetime import datetime
```

```python
con = duckdb.connect(database="dados_duckdb.db", read_only=False)
```

```python
file_path = "../landing/"
# read csv files.
# arquivo = "z0019_1.csv"
arquivo = "z0019_2.csv"
data_ingestao = datetime.now()
df = pd.read_csv(f"{file_path}{arquivo}", sep=";")
df['nome_arquivo'] = arquivo
df['data_ingestao'] = data_ingestao
# head() === limita a quantidade de linhas que serão exibidas
df.head()
```

```python
# ingestao na tabela bronze
con.execute("""
create table if not exists bronze_produtos(
NATBR VARCHAR,
MAKTX VARCHAR,
WERKS VARCHAR,
MAINS VARCHAR,
LABST VARCHAR,
nome_arquivo VARCHAR,
data_ingestao TIMESTAMP
)
""")
```

```python
con.execute("INSERT INTO bronze_z0019 SELECT * FROM df")
```

```python
result = con.execute("SELECT * FROM bronze_z0019").fetchdf()
print(result)
```
