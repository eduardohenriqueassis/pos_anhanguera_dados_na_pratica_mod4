```python
import duckdb
```

```python
con = duckdb.connect(database='dados_duckdb.db', read_only=False)
```

```python
df = con.execute("SELECT * FROM produtos").fetch_df()
df.head(10)
```

```python
# criar uma dimensão de categorias
df2 = df.drop(columns=['id_categoria', 'id_fornecedor'])
df2.head(10)
```

```python
# criar tabela e inserir dados
con.execute("""
    CREATE TABLE IF NOT EXISTS dim_produtos (
        id_produto BIGINT,
        nm_produto TEXT,
        vl_unitario FLOAT
    )
""")
```

```python
con.execute("""
    INSERT INTO dim_produtos select * from df2
""")
```

```python
df_dim = con.execute("SELECT * FROM dim_produtos").fetch_df()
df_dim.head(10)

```

```python
con.close()
```
